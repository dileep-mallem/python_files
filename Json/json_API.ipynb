**JSOn WIth APIs**

```python
# API is: An Application Programming Interface
#  is just a way for two programs to talk to each other over the 
# internet (like a waiter taking your order to the kitchen and bringing back food)

# Status Codes: Servers reply with numbers to show how it went. 
# 200 means OK/success; 4xx or 5xx numbers mean an error happened

# Content-Type: application/json describes a request body; 
#  Accept: application/json states preferred response type
#  With requests, GET then raise_for_status() and response.json()
# json() parses the body but does not check success by itself
#  POST with json=payload serializes the body and sets the JSON content type.

# Python Application
#        |
#        | HTTP Request
#        |
#        v
#    Weather API
#        |
#        | Processes Request
#        |
#        v
#    Weather Server
#        |
#        | JSON Response
#        |
#        v
# Python Application
#        |
#        | Parse JSON
#        |
#        v
# Python Dictionary
#        |
#        v
# Display Weather

# REST API -> Representational State Transfer 
# A REST API exposes resources through URLs and commonly uses HTTP methods.

# Common HTTP methods:
# Method	Purpose	Example
# GET	Retrieve data	Get user information
# POST	Create data	Register a new user
# PUT	Replace a resource	Replace user profile
# PATCH	Partially update	Update user email
# DELETE	Delete data	Delete a user

# https://api.example.com/users 
# GET     /users
# GET     /users/101
# POST    /users
# PUT     /users/101
# PATCH   /users/101
# DELETE  /users/101

# The server may return JSON for these operations.
```

```python
# Making First GET request 
# We'll use JSONPlaceholder , a public fake REST API for learning and testing 

# API Endpoint : https://jsonplaceholder.typicode.com/users/1 

import requests 

url="https://jsonplaceholder.typicode.com/users/1"

response=requests.get(url)

print(response.status_code)  
print(response.text)
```

```python
# 1. requests.get(url) sent an HTTP GET request.
# 2. The server processed it.
# 3. The server returned an HTTP response.
# 4. response.status_code provided the HTTP status code.
# 5. response.text provided the response body as a string.
```

**Understanding HTTP status codes**

```python

# Status codes tell you whether an HTTP request succeeded or failed.
# Code	Meaning
# 200	OK
# 201	Created
# 204	No Content
# 400	Bad Request
# 401	Unauthorized
# 403	Forbidden
# 404	Not Found
# 409	Conflict
# 429	Too Many Requests
# 500	Internal Server Error
# 502	Bad Gateway
# 503	Service Unavailable

# A production application should not assume every request succeeds.
```

**GET request with error Handling**

```python
import requests 

url= "https://jsonplaceholder.typicode.com/users/1"

try : 
    response=requests.get(url,timeout=10)

    response.raise_for_status()

    data=response.json()

    print("Name:",data["name"])
    print('Email:',data["email"])
except requests.exceptions.Timeout : 
    print("The Request Time Out ")

except requests.exceptions.HTTPError as error : 
    print("HTTP errror : ",error)

except requests.exceptions.ConnectionError : 
    print("Could not connect to the Server")
except ValueError : 
    print("THe Server Didnnot return valid JSON")

# timeout=10 prevents waiting indefinitely.
# raise_for_status() raises an exception for unsuccessful HTTP status codes.
# The exception handlers make failures easier to diagnose.
# Invalid JSON is handled separately.

# For a production application, also consider logging errors rather than only printing them
```

**Sending JSON Data to APIs**


POST Request : 
Suppose I want to create a new user
The Server Excepts : 
{
    "name": "Dileep",
    "username": "dileep123",
    "email": "dileep@example.com"
}

```python
url1 = "https://jsonplaceholder.typicode.com/users"
user={
    "name": "Dileep",
    "username": "dileep123",
    "email": "dileep@example.com"
}

response=requests.post(url1,json=user)
print(response.status_code)
print(response.json())

# json= parameter:
# 1. Serializes the Python object into JSON.
# 2. Sends it as the request body.
# 3. Sets the appropriate JSON content type header.
# You do not need to manually call json.dumps() when using json=
```

**Difference between json= and data=**


requests.post(url, json=user) -> The dictionary is serialized as JSON automatically

import json

requests.post(
    url,
    data=json.dumps(user),
    headers={"Content-Type": "application/json"}
)

Here, you manually convert the dictionary into a JSON string and specify the content type.

json=	                        data=
Automatically serializes JSON	Sends the supplied body

**PUT and PATCH Requests**

```python
# PUT replaces a resource. 

updated_user={
    "id":1,
    "name":"Dileep",
    "email":"mailtodileepmallem@gmail.com"
}

response=requests.put(url,json=updated_user)

print(response.status_code)
print(response.json())
```

```python
# PATCH partially updates a resource. 
url = "https://jsonplaceholder.typicode.com/users/1"
updated_data={
    "email":"updated@exaample.com"
}

response=requests.patch(url,json=updated_data)

print(response.status_code)
print(response.json())
```

**API Parameters  Headers and Authentication**

*Query Parameters*

```python
# Real APIs usually require additional information.
# Query parameters are values included in the URL to filter or customize a request.

# https://api.example.com/products?category=laptop&limit=10

url = "https://fakestoreapi.com"

# FakeStoreAPI uses different parameter names, 
# but it allows you to limit results like this:
params = {
    "limit": 10
}

try:
    response = requests.get(url, params=params)
    print(f"Status Code: {response.status_code}")
    print(response.json())
except requests.exceptions.RequestException as e:
    print(f"Connection failed: {e}") 


# Requests builds the query string automatically.
# This is especially useful for search, pagination and filtering.
```

*Headers*


```python
# Headers provide additional information about an HTTP request.

url = "https://api.example.com/data"

headers={
    "Accept" : "application/json",
    "User-Agent":"MyPythonApplication/1.0"
}
response = requests.get(url, headers=headers)

# Common headers:
# Header	Purpose
# Content-Type	Format of the request body
# Accept	Preferred response format
# Authorization	Authentication credentials
# User-Agent	Identifies the client
```

*API Authentication using API Keys*

```python
# Some APIs require an API key.

import requests 
import os 



url = "https://pokeapi.co" # Returns about pikachu
api_key=os.getenv("MY_API_KEY","default_mock_key")

headers={
    "Authorization": f"Bearer {api_key}",
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"

}
print(f"DEBUG: Target URL is: {url}")
print("sendiing Request to PokeAPI with API Key:")

response = requests.get(url,headers=headers)
print(f"Status Code: {response.status_code}")

# print the results 
if response.status_code == 200:
    print("Successfully retrieved JSON data!")
    data = response.json()
    print(f"Name: {data['name'].upper()}")
    print(f"Base Experience: {data['base_experience']}")
else:
    print(f"Failed. Status: {response.status_code}")

```
